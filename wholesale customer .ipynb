```python

# Check for duplicate rows
print("Duplicate rows:", wholesale.duplicated().sum())

# Remove duplicate rows, if any
wholesale = wholesale.drop_duplicates()

# Check missing values
print("\nMissing values in each column:")
print(wholesale.isnull().sum())

# Display the cleaned dataset shape
print("\nDataset shape after cleaning:", wholesale.shape)

# Display the first 5 rows
wholesale.head()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Columns containing customer spending information
spending_columns = [
    "Fresh",
    "Milk",
    "Grocery",
    "Frozen",
    "Detergents_Paper",
    "Delicassen"
]

# Display basic statistics for spending
wholesale[spending_columns].describe()
```

```python

# Plot the spending distribution for each category
wholesale[spending_columns].hist(
    figsize=(14, 9),
    bins=25,
    edgecolor="black"
)

plt.suptitle("Distribution of Wholesale Customer Spending", fontsize=16)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
from sklearn.preprocessing import StandardScaler

# Select the six spending columns
X = wholesale[spending_columns].copy()

# Reduce the effect of very large spending values
X_log = np.log1p(X)

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_log)

print("Original spending data shape:", X.shape)
print("Prepared data shape:", X_scaled.shape)
print("Data preparation completed successfully!")
```

```python

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

scores = {}

for k in range(2, 9):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    scores[k] = score

    print(f"Clusters: {k}, Silhouette Score: {score:.4f}")

# Plot the results
plt.figure(figsize=(8, 5))
plt.plot(list(scores.keys()), list(scores.values()), marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Choosing the Number of Customer Segments")
plt.xticks(list(scores.keys()))
plt.grid(True)
plt.show()
```

```python

# Select the K with the highest silhouette score
best_k = max(scores, key=scores.get)

print("Selected number of clusters:", best_k)

# Train the final K-Means model
kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

wholesale["Cluster"] = kmeans.fit_predict(X_scaled)

# Display the number of customers in each cluster
print("\nCustomers in each cluster:")
print(wholesale["Cluster"].value_counts().sort_index())

# Show some segmented customers
wholesale[["Channel", "Region"] + spending_columns + ["Cluster"]].head(10)
```

```python
import os
import pickle

# Create a project folder
os.makedirs("/content/wholesale_project", exist_ok=True)

# Save the trained model and preprocessing tools
with open("/content/wholesale_project/kmeans_model.pkl", "wb") as f:
    pickle.dump(kmeans, f)

with open("/content/wholesale_project/scaler.pkl", "wb") as f:
    pickle.dump(scaler, f)

with open("/content/wholesale_project/spending_columns.pkl", "wb") as f:
    pickle.dump(spending_columns, f)

# Save customer data with cluster labels
wholesale.to_csv(
    "/content/wholesale_project/customer_segments.csv",
    index=False
)

print("Model and customer data saved successfully!")
```

```python
%%writefile /content/wholesale_project/app.py
import streamlit as st
import pandas as pd
import numpy as np
import pickle

st.set_page_config(page_title="Wholesale Customer Segmentation",
                   page_icon="📊", layout="wide")

folder = "/content/wholesale_project"

with open(f"{folder}/kmeans_model.pkl", "rb") as f:
    model = pickle.load(f)

with open(f"{folder}/scaler.pkl", "rb") as f:
    scaler = pickle.load(f)

with open(f"{folder}/spending_columns.pkl", "rb") as f:
    columns = pickle.load(f)

data = pd.read_csv(f"{folder}/customer_segments.csv")

st.sidebar.title("Wholesale Analytics")
page = st.sidebar.radio(
    "Navigate",
    ["Home", "Dashboard", "Predict Customer", "About"]
)

if page == "Home":
    st.title("📊 Wholesale Customer Segmentation")
    st.write("Understand customer purchasing behavior using machine learning.")
    st.metric("Total Customers", len(data))
    st.metric("Customer Segments", data["Cluster"].nunique())
    st.info("Use the sidebar to explore the dashboard or predict a customer's segment.")

elif page == "Dashboard":
    st.title("Customer Dashboard")

    c1, c2 = st.columns(2)
    c1.metric("Total Customers", len(data))
    c2.metric("Number of Segments", data["Cluster"].nunique())

    st.subheader("Customers in Each Segment")
    st.bar_chart(data["Cluster"].value_counts().sort_index())

    st.subheader("Average Spending by Segment")
    st.dataframe(data.groupby("Cluster")[columns].mean().round(2))

    st.subheader("Customer Data")
    st.dataframe(data, use_container_width=True)

    st.download_button(
        "Download Segmented Customer Data",
        data.to_csv(index=False),
        "customer_segments.csv",
        "text/csv"
    )

elif page == "Predict Customer":
    st.title("Predict a Customer Segment")
    st.write("Enter a customer's annual spending for each category.")

    values = {}
    for col in columns:
        values[col] = st.number_input(
            col, min_value=0.0, value=1000.0, step=100.0
        )

    if st.button("Predict Segment"):
        customer_input = pd.DataFrame([values], columns=columns)
        transformed = scaler.transform(np.log1p(customer_input))
        prediction = model.predict(transformed)[0]
        st.success(f"This customer belongs to Segment {prediction}.")

elif page == "About":
    st.title("About This Project")
    st.write("""
    This project groups wholesale customers based on their spending habits.

    - **Data preparation:** Log transformation and standardization
    - **Machine learning:** K-Means clustering
    - **Dashboard:** Streamlit
    - **Goal:** Identify customers with similar purchasing patterns

    Segment numbers are group labels, not rankings.
    """)
```
